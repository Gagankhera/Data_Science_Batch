## 1. What is Programming?
Programming is the process of writing instructions that a computer can understand and execute.

**Real-world analogy:** A recipe tells a chef how to prepare a dish. A program tells a computer how to perform a task.

## 2. Why Programming Languages?
Computers understand machine language (0s and 1s). Programming languages let humans write instructions in a readable form which are translated for the computer.

## 3. What is Python?
Python is a high-level, interpreted, general-purpose programming language created by **Guido van Rossum** and first released in 1991.

### Why Python?
- Easy syntax
- Large community
- Cross-platform
- Rich libraries
- Excellent for Data Science, AI, Automation and Web Development
