## 1. Introduction to Data Structures

A **data structure** is a way of organizing and storing data so it can be used efficiently.

Python has 4 built-in data structures:

| Structure | Ordered | Mutable | Duplicates Allowed | Syntax |
|---|---|---|---|---|
| List | Yes | Yes | Yes | `[1, 2, 3]` |
| Tuple | Yes | No | Yes | `(1, 2, 3)` |
| Set | No | Yes | No | `{1, 2, 3}` |
| Dictionary | Yes (3.7+) | Yes | Keys: No, Values: Yes | `{"a": 1}` |

Let's explore each one in depth.

# 2. Lists

A **list** is an ordered, mutable collection that can hold items of different data types.


````python
# Creating a list
fruits = ["apple", "banana", "cherry"]
mixed = [1, "hello", 3.14, True]
empty_list = []

print(fruits)
print(mixed)
print(type(fruits))
````

### 2.1 Indexing and Slicing

````python
numbers = [10, 20, 30, 40, 50]

# Indexing (0-based)
print(numbers[0])    # first element -> 10
print(numbers[-1])   # last element  -> 50

# Slicing [start:stop:step]
print(numbers[1:4])   # [20, 30, 40]
print(numbers[:3])    # [10, 20, 30]
print(numbers[::2])   # [10, 30, 50]
print(numbers[::-1])  # reversed list
````

### 2.2 List is Mutable

````python
numbers = [10, 20, 30]
numbers[0] = 100     # modifying an element
print(numbers)

numbers.append(40)   # add to end
print(numbers)
````

### 2.3 Important List Methods

````python
nums = [3, 1, 4, 1, 5, 9, 2]

nums.append(6)          # add single item at end
print("append:", nums)

nums.insert(0, 100)     # insert at index
print("insert:", nums)

nums.remove(1)          # removes FIRST occurrence of value 1
print("remove:", nums)

popped = nums.pop()     # removes & returns last item
print("pop:", nums, "-> popped:", popped)

nums.sort()             # sorts in place (ascending)
print("sort:", nums)

nums.sort(reverse=True) # descending
print("sort desc:", nums)

nums.reverse()          # reverses order in place
print("reverse:", nums)

print("count of 100:", nums.count(100))
print("index of 4:", nums.index(4))

nums.extend([7, 8])     # merge another iterable
print("extend:", nums)

nums2 = nums.copy()     # shallow copy
nums.clear()            # empties the list
print("clear original:", nums)
print("copy kept:", nums2)
````

**Common pitfall:** `list.sort()` sorts in-place and returns `None`. Use `sorted(list)` if you want a **new** sorted list without modifying the original.

````python
original = [3, 1, 2]
new_list = sorted(original)
print("original:", original)
print("new_list:", new_list)
````

### 2.4 List Comprehension

````python
# Basic syntax: [expression for item in iterable if condition]

squares = [x**2 for x in range(1, 6)]
print(squares)

even_numbers = [x for x in range(20) if x % 2 == 0]
print(even_numbers)

# Nested comprehension - flatten a matrix
matrix = [[1, 2, 3], [4, 5, 6]]
flattened = [num for row in matrix for num in row]
print(flattened)
````

### 2.5 Nested Lists (2D Lists)

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[1])       # second row -> [4, 5, 6]
print(matrix[1][2])    # row 1, col 2 -> 6

# Iterating over a nested list
for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
````

### 2.6 Useful Built-in Functions with Lists

````python
nums = [4, 2, 9, 1, 7]

print("len:", len(nums))
print("max:", max(nums))
print("min:", min(nums))
print("sum:", sum(nums))
print("sorted:", sorted(nums))
print("30 in list?:", 30 in nums)
print("2 in list?:", 2 in nums)
````

# 3. Tuples

A **tuple** is an ordered, **immutable** collection. Once created, its elements cannot be changed.


````python
# Creating tuples
coordinates = (10, 20)
single_item = (5,)     # NOTE: comma is required for a single-item tuple
mixed_tuple = (1, "two", 3.0)
without_parens = 1, 2, 3   # parentheses are optional

print(coordinates)
print(single_item, type(single_item))
print(without_parens, type(without_parens))
````

### 3.1 Tuples are Immutable

````python
point = (10, 20)
try:
    point[0] = 100   # this will raise an error
except TypeError as e:
    print("Error:", e)
````

### 3.2 Tuple Methods (only 2 exist, since tuples are immutable)

````python
t = (1, 2, 3, 2, 4, 2)
print("count of 2:", t.count(2))
print("index of 3:", t.index(3))
````

### 3.3 Packing and Unpacking

````python
# Packing
person = ("Alice", 25, "Engineer")

# Unpacking
name, age, profession = person
print(name, age, profession)

# Unpacking with *
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)
````

### 3.4 Why use Tuples over Lists?
- Faster than lists (slightly less overhead).
- Used for data that shouldn't change (e.g., coordinates, RGB values, database records).
- Can be used as **dictionary keys** (lists cannot, since keys must be hashable/immutable).


````python
# Tuple as dictionary key
locations = {
    (28.7041, 77.1025): "Delhi",
    (19.0760, 72.8777): "Mumbai"
}
print(locations[(28.7041, 77.1025)])
````

# 4. Sets

A **set** is an unordered collection of **unique** elements. Duplicates are automatically removed.


````python
fruits = {"apple", "banana", "cherry", "apple"}  # duplicate 'apple' removed
print(fruits)

empty_set = set()          # NOTE: {} creates a dict, not a set!
print(type(empty_set))

nums_set = set([1, 2, 2, 3, 3, 3])
print(nums_set)
````

### 4.1 Sets are Unordered and Unindexed

````python
s = {10, 20, 30}
try:
    print(s[0])   # this will raise an error - sets don't support indexing
except TypeError as e:
    print("Error:", e)
````

### 4.2 Adding and Removing Elements

````python
s = {1, 2, 3}

s.add(4)
print("add:", s)

s.update([5, 6, 7])   # add multiple items
print("update:", s)

s.remove(6)            # raises KeyError if not present
print("remove:", s)

s.discard(100)          # does NOT raise error if not present
print("discard (safe):", s)

popped = s.pop()        # removes an arbitrary element
print("pop:", s, "-> popped:", popped)
````

### 4.3 Mathematical Set Operations

````python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print("Union:", a | b, "or", a.union(b))
print("Intersection:", a & b, "or", a.intersection(b))
print("Difference (a-b):", a - b, "or", a.difference(b))
print("Symmetric Difference:", a ^ b, "or", a.symmetric_difference(b))
````

````python
# Subset / Superset checks
x = {1, 2}
y = {1, 2, 3, 4}

print("x is subset of y?", x.issubset(y))
print("y is superset of x?", y.issuperset(x))
print("x and {9,10} are disjoint?", x.isdisjoint({9, 10}))
````

### 4.4 Why Sets? — Removing Duplicates Fast
A very common real-world use case: removing duplicates from a list.


````python
nums = [1, 2, 2, 3, 4, 4, 4, 5]
unique_nums = list(set(nums))
print(unique_nums)
````

### 4.5 Set Comprehension

````python
squares_set = {x**2 for x in range(1, 6)}
print(squares_set)
````

# 5. Dictionaries

A **dictionary** stores data as **key-value pairs**. Keys must be unique and immutable (hashable); values can be anything.


````python
student = {
    "name": "Ravi",
    "age": 21,
    "branch": "CSE"
}
print(student)
print(type(student))
````

### 5.1 Accessing Values

````python
print(student["name"])          # direct access - KeyError if missing
print(student.get("age"))       # safe access - returns None if missing
print(student.get("grade", "Not Available"))  # default value if key missing
````

### 5.2 Adding, Updating, Removing

````python
student["grade"] = "A"          # add new key
print("after add:", student)

student["age"] = 22             # update existing key
print("after update:", student)

del student["grade"]            # delete a key
print("after del:", student)

popped_value = student.pop("branch")  # remove & return value
print("after pop:", student, "-> popped:", popped_value)
````

### 5.3 Useful Dictionary Methods

````python
student = {"name": "Ravi", "age": 21, "branch": "CSE"}

print("keys:", student.keys())
print("values:", student.values())
print("items:", student.items())

# Looping through a dictionary
for key, value in student.items():
    print(key, "->", value)
````

````python
d1 = {"a": 1, "b": 2}
d2 = {"b": 20, "c": 3}

d1.update(d2)     # merges d2 into d1 (overwrites shared keys)
print(d1)

# Python 3.9+ merge operator
merged = {"a": 1, "b": 2} | {"b": 20, "c": 3}
print(merged)
````

### 5.4 Dictionary Comprehension

````python
squares_dict = {x: x**2 for x in range(1, 6)}
print(squares_dict)

# with condition
even_squares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(even_squares)
````

### 5.5 Nested Dictionaries

````python
students = {
    "s1": {"name": "Ravi", "marks": 85},
    "s2": {"name": "Priya", "marks": 92}
}

print(students["s2"]["name"])

for student_id, info in students.items():
    print(f"{student_id}: {info['name']} scored {info['marks']}")
````

### 5.6 `in` keyword with Dictionaries — checks KEYS by default

````python
student = {"name": "Ravi", "age": 21}
print("name" in student)     # True (checks keys)
print("Ravi" in student)     # False (does NOT check values)
print("Ravi" in student.values())  # True - check values explicitly
````

# 6. Comparison Table — When to Use What?

| Use Case | Best Choice | Why |
|---|---|---|
| Data that changes often, ordered | **List** | Mutable, supports indexing |
| Fixed data that shouldn't change | **Tuple** | Immutable, safer, slightly faster |
| Removing duplicates / membership tests | **Set** | Automatically unique, very fast `in` checks |
| Key-value / lookup data | **Dictionary** | Fast lookup by key, human-readable |


````python
import timeit

big_list = list(range(100000))
big_set = set(range(100000))

list_time = timeit.timeit("99999 in big_list", globals=globals(), number=100)
set_time = timeit.timeit("99999 in big_set", globals=globals(), number=100)

print(f"List membership check time: {list_time:.6f} sec")
print(f"Set membership check time:  {set_time:.6f} sec")
print("Sets are dramatically faster for membership testing!")
````
