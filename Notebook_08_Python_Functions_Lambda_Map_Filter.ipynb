# Functions, Lambda, Map and Filter in Python


## Getting Started With Functions
- What is a function?
- A function is a reusable block of code that performs a specific task.

Keywords: reusable, block of code, specific task

- Syntax
- Parameters & Arguments
- Return statement

### Function → A reusable block of code that performs a specific task.
### Parameter → The input given to the function.
### Argument → The actual value passed to the function.
### print() → Displays the result on the screen only.
### return → Sends the result back to the caller so it can be stored, reused, or used in further calculations.

**Example:** Basic function

```python
def greet():
    return f"Good Morning!"
print(greet())
```

```python
def greet(name):
    return f'Hello {name}'

print(greet('Alice'))
```

**Example:** Addition

```python
def add(a,b):
    return a+b
print(add(5,7))
```

## More Coding Examples With Functions


**Example:** Factorial

```python
# Using builtin function
import math
print(math.factorial(5))
```

```python
def factorial(n):
    f=1
    for i in range(1,n+1):
        f = f*i
    return f
print(factorial(5))
```

**Example:** Even check

```python
def is_even(n):
    return n%2==0
print(is_even(8))
```

## Lambda Function In Python


A lambda function is a small, anonymous (nameless) function that is used for short, one-line operations.

```python
square=lambda x:x*x
print(square(6))
```

```python
maximum=lambda a,b:a if a>b else b
print(maximum(4,9))
```

## Map Function In Python


The map() function applies the same operation to every element in an iterable and returns a new iterable.

```python
nums=[1,2,3,4]

r = []
for i in nums:
    r.append(i*i)
print(r)
```

```python
nums=[1,2,3,4]
print(list(map(lambda x:x*x,nums)))
```

## Filter Function In Python


#### filter() selects only those elements that satisfy a given condition.

#### filter() removes unwanted data and keeps only the required data.

**Example:** Filter evens

```python
nums=[1,2,3,4,5,6]
print(list(filter(lambda x:x%2==0,nums)))
```
