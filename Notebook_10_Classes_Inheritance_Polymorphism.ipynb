
### 📖 Explanation

- A **class** is a blueprint (template) for creating objects. It defines *attributes* (data) and *methods* (functions/behaviour) that describe what the object looks like and what it can do.
- An **object** is a specific *instance* of a class — created using that blueprint, with its own actual values.


Think of a class as a "Car" design drawing, and objects as the actual cars (Honda City, Maruti Swift, etc.) built from that design.

Key building blocks:
- `class` keyword — defines a class.
- `__init__()` — a special method (constructor) that runs automatically when an object is created. It's used to initialize attributes.
- `self` — refers to the current object instance; it must be the first parameter of every instance method.
- **Attributes** — variables that belong to an object (e.g. `self.name`).
- **Methods** — functions defined inside a class that describe the object's behaviour.

### Syntax
```python
class ClassName:
    def __init__(self, param1, param2):
        self.attribute1 = param1
        self.attribute2 = param2

    def some_method(self):
        # behaviour using self.attribute1, self.attribute2
        pass

obj = ClassName(value1, value2)   # creating an object
```


````python

# Simple example: A Class and Object representing a Student

class Student:
    # constructor - runs automatically when a new object is created
    def __init__(self, name, age, grade):
        self.name = name        # attribute
        self.age = age          # attribute
        self.grade = grade      # attribute

    # a method (behaviour)
    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}, Grade: {self.grade}")

    def is_adult(self):
        return self.age >= 18


# Creating objects (instances) of the Student class
student1 = Student("Aditi", 16, "10th")
student2 = Student("Rohan", 19, "12th")

# Using the objects
student1.display_info()
student2.display_info()

print(f"Is {student1.name} an adult? {student1.is_adult()}")
print(f"Is {student2.name} an adult? {student2.is_adult()}")
````

Instead of writing the same code repeatedly, we inherit it

````python
class Vehicle:

    def start(self):
        print("Vehicle Started")

    def stop(self):
        print("Vehicle Stopped")
````

````python
# Parent Class
class Vehicle:

    def start(self):
        print("Vehicle Started")

    def stop(self):
        print("Vehicle Stopped")


# Child Class
class Car(Vehicle):

    def music(self):
        print("Music is Playing")



# Creating Object
car = Car()

car.start()     # Inherited from Vehicle
car.music()     # Car's own method
car.stop()      # Inherited from Vehicle'''
````


### 📖 Explanation

**Inheritance** allows one class (the **child/derived class**) to reuse and extend the attributes and methods of another class (the **parent/base class**). This promotes code reuse and creates a natural "is-a" relationship.

For example: A `Dog` **is an** `Animal`. A `Car` **is a** `Vehicle`.

Key terms:
- **Parent class (Base class)** — the class being inherited from.
- **Child class (Derived class)** — the class that inherits.
- **`super()`** — used inside the child class to call the parent class's methods (like its `__init__`).
- **Method Overriding** — a child class can redefine a method from the parent to change its behaviour.

Types of inheritance in Python:
1. Single Inheritance (one parent, one child)
2. Multiple Inheritance (child inherits from more than one parent)
3. Multilevel Inheritance (grandparent → parent → child)
4. Hierarchical Inheritance (one parent, multiple children)

### Syntax
```python
class Parent:
    def __init__(self, x):
        self.x = x

class Child(Parent):
    def __init__(self, x, y):
        super().__init__(x)   # calls Parent's __init__
        self.y = y
```


Why super().__init__(brand)?
Vehicle has a constructor (__init__) that initializes brand.
Car also has its own constructor for model.
super().__init__(brand) calls the parent class constructor so that brand is initialized correctly.

Example 2: Child Class with Additional Attributes

````python
class Vehicle:

    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(f"{self.brand} Vehicle Started")

    def stop(self):
        print(f"{self.brand} Vehicle Stopped")


class Car(Vehicle):

    def __init__(self, brand, model):
        super().__init__(brand)   # Calls Vehicle's constructor
        self.model = model

    def display(self):
        print("Brand :", self.brand)
        print("Model :", self.model)


car = Car("Toyota", "Fortuner")

car.display()
car.start()
car.stop()
````

````python

# Simple example: Inheritance with Animal -> Dog

class Animal:
    def __init__(self, name):
        self.name = name

    def make_sound(self):
        print(f"{self.name} makes a generic animal sound.")

    def eat(self):
        print(f"{self.name} is eating.")


# Dog inherits from Animal
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)     # reuse Animal's constructor
        self.breed = breed

    # method overriding - Dog has its own version of make_sound
    def make_sound(self):
        print(f"{self.name} (a {self.breed}) barks: Woof! Woof!")


generic_animal = Animal("Some Animal")
generic_animal.make_sound()

my_dog = Dog("Tommy", "Labrador")
my_dog.make_sound()   # uses the overridden method
my_dog.eat()           # inherited method from Animal
````

### 📖 Explanation

**Polymorphism** means "many forms." In OOP, it refers to the ability of different objects to respond to the *same method call* in different ways.

Common forms of polymorphism in Python:

1. **Method Overriding (Runtime Polymorphism)** — Different classes have a method with the same name, but each class implements it differently. When you call that method, Python decides which version to run based on the actual object type.

2. **Duck Typing** — Python does not check the type of an object before calling a method; if the object has the method, it can be used. "If it walks like a duck and quacks like a duck, it's a duck."

3. **Built-in Polymorphism** — Many built-in functions/operators behave differently based on the data type, e.g. `+` adds numbers but concatenates strings; `len()` works on strings, lists, dictionaries, etc.

### Why is it useful?
Polymorphism lets you write generic code that works with objects of different classes, as long as they support the expected methods — making code flexible and extensible.


````python

# Simple example: Polymorphism with different Shape classes

class Shape:
    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


# The SAME loop calls area() on different object types -> polymorphism
shapes = [Circle(5), Square(4), Triangle(6, 3)]

for shape in shapes:
    print(f"{shape.__class__.__name__} area: {shape.area()}")

# Built-in polymorphism example: + behaves differently based on type # Operator overloading
print(3 + 5)          # addition
print("Py" + "thon")  # concatenation
````
