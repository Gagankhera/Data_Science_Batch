## 1. Why Do We Need Loops?

Imagine printing "Hello" 5 times without a loop:

```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```

That's repetitive and doesn't scale (what if you needed it 1000 times?). A loop solves this by repeating a block of code automatically.


````python
# Without a loop
print("Hello")
print("Hello")
print("Hello")
````

````python
# With a loop
for i in range(3):
    print("Hello")
````

## 2. The `for` Loop

A `for` loop is used to iterate over a **sequence** (like a range of numbers, a string, or a list), running the indented block once for each item.

### Syntax
```python
for item in sequence:
    # code block - runs once per item
```


````python
# Looping over a range of numbers
for i in range(5):
    print(i)
````

### The `range()` function

`range()` generates a sequence of numbers and is the most common way to control how many times a `for` loop runs.

| Form | Meaning |
|---|---|
| `range(stop)` | 0, 1, 2, ... up to (but not including) `stop` |
| `range(start, stop)` | `start` up to (not including) `stop` |
| `range(start, stop, step)` | `start` up to `stop`, incrementing by `step` |


````python
print("range(5):", list(range(5)))
print("range(2, 6):", list(range(2, 6)))
print("range(0, 10, 2):", list(range(0, 10, 2)))
print("range(10, 0, -1):", list(range(10, 0, -1)))
````

## 3. Looping Over Strings and Lists

A `for` loop can directly iterate over the characters of a string, or the elements of a list, without needing `range()`.


````python
# Looping over a string
for ch in "Python":
    print(ch)
````

````python
# Looping over a list
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

## 4. The `while` Loop

A `while` loop repeats a block of code **as long as** a condition remains `True`. It is useful when you don't know in advance exactly how many times you need to loop.

### Syntax
```python
while condition:
    # code block - runs while condition is True
```

> ⚠️ Make sure the condition eventually becomes `False`, otherwise you get an **infinite loop**!


````python
# Basic while loop
count = 1

while count <= 5:
    print(count)
    count += 1   # this MUST change, or the loop never ends
````

````python
# while loop counting down
n = 5

while n > 0:
    print(n)
    n -= 1

print("Liftoff! 🚀")
````

## 5. `break` and `continue`

- **`break`** — immediately exits the loop, skipping any remaining iterations.
- **`continue`** — skips the rest of the current iteration and moves to the next one.


````python
# break example: stop as soon as we find the number 5
for i in range(1, 10):
    if i == 5:
        print("Found 5, stopping the loop.")
        break
    print(i)
````

````python
# continue example: skip even numbers
for i in range(1, 10):
    if i % 2 == 0:
        continue
    print(i)
````

## 6. The `pass` Statement

`pass` does nothing — it's a placeholder used when a statement is syntactically required but you don't want to write any code yet.


````python
for i in range(5):
    if i == 3:
        pass   # TODO: handle this case later
    else:
        print(i)
````

## 7. Nested Loops

A loop can be placed inside another loop. The inner loop completes all its iterations for **each single** iteration of the outer loop.


````python
# Nested loop: multiplication table for 1 to 3
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i*j}")
    print("---")
````

## 8. Quick Recap

| Concept | Syntax |
|---|---|
| for loop | `for item in sequence:` |
| range | `range(stop)`, `range(start, stop)`, `range(start, stop, step)` |
| while loop | `while condition:` |
| break | exits the loop immediately |
| continue | skips to the next iteration |
| pass | does nothing (placeholder) |
| nested loop | a loop inside another loop |

**Golden Rule:** Always make sure a `while` loop's condition will eventually become `False` — otherwise your program runs forever!


## 9. Practical Examples

Let's apply loops to a few simple, everyday programs.


````python
# Example 1: Sum of first N natural numbers
n = 10
total = 0

for i in range(1, n + 1):
    total += i

print(f"Sum of first {n} natural numbers = {total}")
````

````python
# Example 2: Multiplication table of a number
num = 7

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
````

````python
# Example 3: Count vowels in a word
word = "Programming"
vowel_count = 0

for ch in word.lower():
    if ch in "aeiou":
        vowel_count += 1

print(f"Number of vowels in '{word}': {vowel_count}")
````

````python
# Example 4: Reverse a string using a while loop
text = "python"
reversed_text = ""
index = len(text) - 1

while index >= 0:
    reversed_text += text[index]
    index -= 1

print(f"Original: {text}")
print(f"Reversed: {reversed_text}")
````

````python
# Example 5: Simple guess-the-number style loop (search using break)
numbers = [4, 8, 15, 16, 23, 42]
target = 16

for num in numbers:
    if num == target:
        print(f"{target} found in the list!")
        break
else:
    # this 'else' belongs to the for loop - runs only if break was never hit
    print(f"{target} not found in the list.")
````
