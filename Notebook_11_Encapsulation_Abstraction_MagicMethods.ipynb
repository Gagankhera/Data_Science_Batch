## 1. Encapsulation in OOPS

---

Wrapping Data and Methods Together
Definition

Encapsulation means:

Keeping data and the methods that operate on that data together in one class, while controlling access to that data.

Simple definition:

"Protect the data from direct access."


### 📖 Explanation

**Encapsulation** is the bundling of data (attributes) and the methods that operate on that data into a single unit (the class), while restricting direct access to some of the object's components. This protects the internal state of an object from unintended or invalid modification.

Python does not have strict "private" access like Java, but it uses **naming conventions**:

| Convention | Example | Meaning |
|---|---|---|
| Public | `self.name` | Accessible from anywhere |
| Protected (convention only) | `self._name` | Should be treated as internal; accessible but "please don't touch outside" |
| Private (name-mangled) | `self.__name` | Python renames it internally to `_ClassName__name` to prevent accidental access from outside |

**Getters and setters** (often implemented using the `@property` decorator) are commonly used to control how private attributes are read or modified, allowing validation logic to be added.

### Why use encapsulation?
- Protects data from being changed accidentally or incorrectly.
- Allows validation before setting a value.
- Hides internal implementation details from the outside world.


Quick comparison:
 
    	           Protected (_var)	                         Private (__var)
    Symbol	      Single underscore	                       Double underscore
                 Accessible in subclass?   	              Yes	Not directly (mangled)
    Accessible 
    outside 
    class?	     Technically yes (just convention)   	  Technically yes, but discouraged/harder
    
    Enforcement	    None — just a hint	                  Name mangling adds a mild barrier
    Use case	"Internal, but subclasses may need it"	 "Truly internal, hide implementation details"

Key mindset: Python doesn't have true private/protected like Java or C++ (which throw compiler errors if you break the rules). Python trusts developers and uses naming conventions instead of hard restrictions — this is often called "we're all consenting adults here."

That's also why @property is so handy in Python — it lets you keep the internal variable private/protected (self.__radius or self._radius) while still giving controlled, clean access from outside.

### Getter and setter methods

### @property decorator

````python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")


s = Student(85)

print(s.get_marks())

s.set_marks(95)

print(s.get_marks())

s.set_marks(150)
````

### With @Property decorator

##### @property is a way to make a method behave like an attribute — you call it without parentheses, but it can still run code behind the scenes.

````python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    @property
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")


s = Student(80)

print(s.marks)       # getter
s.marks = -90       # setter
 
````

````python

# Simple example: Encapsulation with a BankAccount class

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # public attribute
        self.__balance = balance    # private attribute (name-mangled)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance!")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")

    # a "getter" method to safely access the private attribute
    def get_balance(self):
        return self.__balance


account = BankAccount("Meera", 1000)
account.deposit(500)
account.withdraw(2000)   # will fail - insufficient balance
account.withdraw(300)
print("Final balance:", account.get_balance())

# Trying to access the private attribute directly will fail:
try:
    print(account.__balance)
except AttributeError as e:
    print("Error:", e)
````

## 2. Abstraction in OOPS

---


### 📖 Explanation

**Abstraction** means hiding complex implementation details and showing only the essential features of an object. The user interacts with a simple interface without needing to know how it works internally.

Example: When you drive a car, you use the steering wheel and pedals (the interface) without knowing exactly how the engine or transmission works internally (the implementation).

In Python, abstraction is commonly implemented using the `abc` module (**Abstract Base Classes**):

- `ABC` — a base class used to create abstract classes.
- `@abstractmethod` — a decorator that marks a method as abstract, meaning it **must** be implemented by any subclass. You cannot create an object directly from a class that has abstract methods.

### Syntax
```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass    # no implementation here - forces subclasses to implement it
```

### Abstraction vs Encapsulation
- **Encapsulation** hides the *data* (restricting access to attributes).
- **Abstraction** hides the *implementation complexity* (showing only relevant methods, forcing a consistent interface).


    Why Use @abstractmethod?
    Ensures all child classes implement required methods.
    Creates a common structure for related classes.
    Prevents incomplete implementations.
    Makes large projects easier to maintain.

| Abstraction                                       | Encapsulation                                                           |
| ------------------------------------------------- | ----------------------------------------------------------------------- |
| Hides implementation details                      | Hides and protects data                                                 |
| Focuses on **what** an object does                | Focuses on **how** data is accessed                                     |
| Simplifies usage                                  | Prevents unauthorized modification                                      |
| Achieved using abstract classes/interfaces        | Achieved using classes and access control (`_` and `__`)                |
| Example: Driving a car without knowing the engine | Example: Accessing a bank account only through deposit/withdraw methods |


Imagine a teacher tells students:

"Every student must submit a project."

The teacher doesn't specify the project itself.

Different students submit different projects:

Student A → Science Project
Student B → Robotics Project
Student C → AI Project

The teacher only makes the rule.

Similarly,

The parent (abstract) class only defines that a method must exist, while each child class decides how to implement it.

````python

# Simple example: Abstraction with Payment methods

from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass    # subclasses MUST implement this method

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card.")

class UpiPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using UPI.")


# Trying to create an object of the abstract class directly will FAIL:
try:
    p = Payment()
except TypeError as e:
    print("Error:", e)

# But subclasses that implement the abstract method work fine:
credit = CreditCardPayment()
upi = UpiPayment()

credit.pay(500)
upi.pay(200)
````

## 3. Magic Methods in Python

---


### 📖 Explanation

**Magic methods** (also called **dunder methods** — "double underscore" methods) are special methods in Python that start and end with double underscores, like `__init__`, `__str__`, `__len__`, `__add__`, etc. They let you define how objects of your class behave with built-in Python operations, such as printing, comparison, arithmetic, and more.

Some common magic methods:

| Magic Method | Purpose |
|---|---|
| `__init__(self, ...)` | Constructor — initializes a new object |
| `__str__(self)` | Defines the "nice/readable" string shown by `print(obj)` or `str(obj)` |
| `__repr__(self)` | Defines the "official" string representation, useful for debugging |
| `__len__(self)` | Defines behaviour for `len(obj)` |
| `__eq__(self, other)` | Defines behaviour for `obj1 == obj2` |
| `__add__(self, other)` | Defines behaviour for `obj1 + obj2` |
| `__getitem__(self, key)` | Defines behaviour for `obj[key]` |

Magic methods are the foundation of **operator overloading** (covered in the next notebook) and let custom objects integrate naturally with Python's built-in syntax.


````python

# Simple example: Magic Methods with a Point class

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # called by print(obj) and str(obj)
    def __str__(self):
        return f"Point({self.x}, {self.y})"

    # called by len(obj) -- here we define it as distance-ish "size" for demo
    def __len__(self):
        return int((self.x ** 2 + self.y ** 2) ** 0.5)

    # called when using obj1 == obj2
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    # called when using obj1 + obj2
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)


p1 = Point(3, 4)
p2 = Point(3, 4)
p3 = Point(1, 2)

print(p1)                 # uses __str__ -> Point(3, 4)
print(len(p1))             # uses __len__ -> 5 (3-4-5 triangle)
print(p1 == p2)            # uses __eq__ -> True
print(p1 == p3)            # uses __eq__ -> False

p4 = p1 + p3               # uses __add__
print(p4)                  # Point(4, 6)
````
