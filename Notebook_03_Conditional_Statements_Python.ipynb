## 1. What is a Conditional Statement?

In real life, we constantly make decisions based on conditions:

> "**If** it is raining, **then** I will take an umbrella, **else** I will wear sunglasses."

Programs need to do the same thing — execute different blocks of code depending on whether a condition is `True` or `False`.

In Python, this is done using:

| Keyword | Meaning |
|---|---|
| `if` | Runs a block of code **only if** a condition is `True` |
| `elif` | Short for "else if" — checks another condition if the previous one(s) were `False` |
| `else` | Runs a block of code if **none** of the above conditions were `True` |

### Key Rule: Indentation
Python does **not** use curly braces `{}` like C, Java, or JavaScript. Instead, it uses **indentation** (usually 4 spaces) to define a block of code. Incorrect indentation is the #1 source of errors for beginners!


````python
# A simple real-world style decision in Python
is_raining = True

if is_raining:
    print("Take an umbrella! ☔")
else:
    print("Wear sunglasses! 😎")
````

## 2. The `if` Statement

### Syntax
```python
if condition:
    # code block - runs only if condition is True
```

- `condition` is any expression that evaluates to `True` or `False` (a boolean).
- The colon `:` at the end is mandatory.
- The indented block below runs **only** when the condition is `True`. If it's `False`, Python simply skips the block.


````python
age = 20

if age >= 18:
    print("You are eligible to vote.")

print("This line always runs, regardless of the condition.")
````

````python
# Try changing the value of 'age' below and re-run the cell
age = 15

if age >= 18:
    print("You are eligible to vote.")

print("Program finished.")
````

## 3. The `if-else` Statement

What if we want to do something *specific* when the condition is `False`, instead of just skipping it? That's where `else` comes in.

### Syntax
```python
if condition:
    # runs if condition is True
else:
    # runs if condition is False
```


````python
number = 7

if number % 2 == 0:
    print(f"{number} is Even")
else:
    print(f"{number} is Odd")
````

## 4. The `if-elif-else` Ladder

When there are **more than two** possibilities, we use `elif`. Python checks each condition from top to bottom and executes the **first** one that is `True`. If none match, the `else` block runs (if present).

### Syntax
```python
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False AND condition2 is True
elif condition3:
    # runs if condition1, condition2 are False AND condition3 is True
else:
    # runs if none of the above are True
```

> ⚠️ **Important:** Only ONE block in the entire if-elif-else ladder executes — the first one whose condition is `True`.


````python
marks = 76

if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 40:
    grade = "C"
else:
    grade = "Fail"

print(f"Marks: {marks} -> Grade: {grade}")
````

## 5. Nested Conditional Statements

An `if` statement can contain another `if` statement inside it. This is called **nesting**, and it's useful when a decision depends on multiple, layered conditions.


````python
username = "admin"
password = "python123"

if username == "admin":
    if password == "python123":
        print("Login Successful ✅")
    else:
        print("Incorrect Password ❌")
else:
    print("Unknown Username ❌")
````

> 💡 **Tip:** Deeply nested `if` statements can get hard to read. Often, combining conditions with logical operators (`and`, `or`) is cleaner — see Section 7.

## 6. Short-Hand `if` and Ternary Operator

### 6.1 Single-line `if` (when there's only one statement)
```python
if condition: print("statement")
```

### 6.2 Ternary (Conditional) Expression
Python offers a compact way to assign a value based on a condition, all in one line:

```python
value = true_result if condition else false_result
```

This is extremely popular for short, simple decisions.


````python
# Single line if
age = 20
if age >= 18: print("Adult")

# Ternary operator example
age = 16
status = "Adult" if age >= 18 else "Minor"
print(status)
````

## 7. Logical Operators with Conditionals

Sometimes a decision depends on **multiple conditions at once**. Python provides three logical operators:

| Operator | Meaning | Example |
|---|---|---|
| `and` | True only if **both** conditions are True | `age >= 18 and has_id` |
| `or` | True if **at least one** condition is True | `is_weekend or is_holiday` |
| `not` | Reverses the boolean value | `not is_raining` |


````python
age = 25
has_id_card = True

if age >= 18 and has_id_card:
    print("Entry allowed to the concert 🎫")
else:
    print("Entry denied")

# 'or' example
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend! 🎉")
else:
    print("It's a weekday. Back to work!")

# 'not' example
is_raining = False
if not is_raining:
    print("Let's go for a walk!")
````

## 8. Common Mistakes to Avoid

1. **Using `=` instead of `==`** — `=` is assignment, `==` is comparison.
2. **Forgetting the colon `:`** at the end of `if`/`elif`/`else`.
3. **Incorrect indentation** — mixing tabs and spaces, or inconsistent spacing.
4. **Unreachable `elif`/`else` after a very broad condition** — order your conditions from most specific to most general.
5. **Comparing floating point numbers with `==`** — due to precision issues, prefer a tolerance check (e.g. `abs(a - b) < 0.0001`).

Let's see mistake #1 and #4 in action:


````python
# Mistake type 4: wrong ordering of conditions (logic bug, not a crash)
marks = 95

# WRONG ORDER - this will incorrectly classify a 95 as grade "C" or worse
if marks >= 40:
    grade = "C or better (but too vague!)"
elif marks >= 90:
    grade = "A+"   # This will NEVER be reached because the first condition already caught it!
else:
    grade = "Fail"

print("Buggy result:", grade)

print("-" * 40)

# CORRECT ORDER - always check the strictest / highest condition first
if marks >= 90:
    grade = "A+"
elif marks >= 40:
    grade = "C or better"
else:
    grade = "Fail"

print("Correct result:", grade)
````

## 9. Practical Examples — Basic Level

Let's now apply everything to simple, self-contained beginner programs.


````python
# Example 1: Check if a number is positive, negative, or zero
num = -4

if num > 0:
    print(f"{num} is Positive")
elif num < 0:
    print(f"{num} is Negative")
else:
    print(f"{num} is Zero")
````

````python
# Example 2: Find the largest of three numbers
a, b, c = 12, 45, 30

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print(f"The largest number among {a}, {b}, {c} is {largest}")
````

````python
# Example 3: Simple eligibility checker
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote.")
else:
    print(f"You need to wait {18 - age} more year(s) to vote.")
````

## 10. Practical Examples — Intermediate Level

````python
# Example 4: Leap Year Checker
year = 2024

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is NOT a Leap Year")
````

````python
# Example 5: Simple Calculator using if-elif-else
num1 = 15
num2 = 4
operator = "/"

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Division by zero!"
else:
    result = "Error: Invalid operator"

print(f"Result: {result}")
````

````python
# Example 6: Password Strength Checker
password = "Py@2024"

length_ok = len(password) >= 6
has_digit = any(ch.isdigit() for ch in password)
has_upper = any(ch.isupper() for ch in password)
has_special = any(ch in "!@#$%^&*" for ch in password)

if length_ok and has_digit and has_upper and has_special:
    print("Strong Password ✅")
elif length_ok and (has_digit or has_upper):
    print("Moderate Password ⚠️")
else:
    print("Weak Password ❌")
````

## 11. Practical Examples — Advanced Level

````python
# Example 7: BMI Calculator with health category (nested + elif ladder)
weight_kg = 68     # in kilograms
height_m = 1.72    # in meters

bmi = weight_kg / (height_m ** 2)
bmi = round(bmi, 2)

if bmi < 18.5:
    category = "Underweight"
elif 18.5 <= bmi < 25:
    category = "Normal weight"
elif 25 <= bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"BMI: {bmi} -> Category: {category}")
````

````python
# Example 8: Income Tax Calculator using slabs (elif ladder with cumulative logic)
income = 950000  # annual income in INR

if income <= 250000:
    tax = 0
elif income <= 500000:
    tax = (income - 250000) * 0.05
elif income <= 1000000:
    tax = (250000 * 0.05) + (income - 500000) * 0.20
else:
    tax = (250000 * 0.05) + (500000 * 0.20) + (income - 1000000) * 0.30

print(f"Income: ₹{income} -> Tax payable: ₹{round(tax, 2)}")
````

````python
# Example 9: Menu-Driven Program (simulates a real console app using nested conditionals)
def restaurant_menu(choice, quantity):
    menu_prices = {1: ("Burger", 150), 2: ("Pizza", 350), 3: ("Pasta", 250), 4: ("Coffee", 80)}

    if choice in menu_prices:
        item_name, price = menu_prices[choice]
        if quantity > 0:
            total = price * quantity
            if total > 500:
                total = total * 0.9  # 10% discount on bills above 500
                print(f"{quantity} x {item_name} = ₹{price*quantity} -> 10% discount applied -> Final: ₹{round(total,2)}")
            else:
                print(f"{quantity} x {item_name} = ₹{total}")
        else:
            print("Quantity must be greater than zero.")
    else:
        print("Invalid menu choice. Please select between 1-4.")

# Simulate a few orders
restaurant_menu(2, 2)   # 2 Pizzas
restaurant_menu(4, 1)   # 1 Coffee
restaurant_menu(9, 1)   # invalid choice
````

````python
# Example 10: Grading System with multiple subjects (combining loops + conditionals - a step towards advanced programs)
students = {
    "Aarav": 88,
    "Diya": 45,
    "Kabir": 72,
    "Meera": 91,
    "Zoya": 33
}

for name, score in students.items():
    if score >= 90:
        grade = "A+"
    elif score >= 75:
        grade = "A"
    elif score >= 60:
        grade = "B"
    elif score >= 40:
        grade = "C"
    else:
        grade = "Fail"
    print(f"{name}: {score} marks -> Grade {grade}")
````

## 12. Quick Recap / Cheat Sheet

| Concept | Syntax |
|---|---|
| Basic if | `if cond:` |
| if-else | `if cond: ... else: ...` |
| if-elif-else | `if c1: ... elif c2: ... else: ...` |
| Nested if | `if c1:` then indented `if c2:` inside |
| Ternary | `x if cond else y` |
| Logical AND | `cond1 and cond2` |
| Logical OR | `cond1 or cond2` |
| Logical NOT | `not cond` |

**Golden Rules:**
1. Always use `==` for comparison, `=` for assignment.
2. Don't forget the colon `:`.
3. Be consistent with indentation (4 spaces recommended).
4. Order your `elif` conditions from most specific/strict to most general.

````python
# Solution 5: Login System (Advanced)
correct_username = "student"
correct_password = "learnpython"
attempts = 0
max_attempts = 3
logged_in = False

while attempts < max_attempts:
    entered_username = input("Username: ")
    entered_password = input("Password: ")

    if entered_username == correct_username and entered_password == correct_password:
        print("Login Successful ✅")
        logged_in = True
        break
    else:
        attempts += 1
        remaining = max_attempts - attempts
        if remaining > 0:
            print(f"Incorrect credentials. Attempts remaining: {remaining}")
        else:
            print("Account locked due to too many failed attempts. 🔒")

if not logged_in and attempts >= max_attempts:
    print("Access denied.")
````
