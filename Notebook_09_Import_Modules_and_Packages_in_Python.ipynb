## What is a Module?
A module is a Python file containing functions, variables, or classes.

````python
import math
print(math.sqrt(25))
print(math.factorial(5))
````

## Different Import Styles

````python
import math
from math import sqrt
from math import pi as PI

print(math.ceil(4.2))
print(sqrt(81))
print(PI)
````

## Creating Your Own Module
Create `calculator.py`:
```python
def add(a,b):
    return a+b
```
Then:
```python
import calculator
print(calculator.add(2,3))
```

## Packages
A package is a folder containing Python modules (often with `__init__.py`).

## Standard Library Overview

````python
import random, datetime, os, statistics

print(random.randint(1,10))
print(datetime.date.today())
print(os.getcwd())
print(statistics.mean([10,20,30]))
````

## Best Practices
- Import only what you need
- Avoid `from module import *`
- Use aliases like `import numpy as np`
