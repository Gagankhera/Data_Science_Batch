## 1️⃣ Tuples

A **tuple** is just like a list — it stores items in order — but with one BIG difference:

🔒 **Tuples cannot be changed after creation.** (This is called being "immutable".)

- Written with **round brackets** `( )`
- Items are separated by commas
- Once created, you **cannot** add, remove, or change items


````python
# Creating a tuple
fruits = ("apple", "banana", "cherry")
print(fruits)
print(type(fruits))
````

````python
# A tuple with just ONE item needs a trailing comma!
single_item = ("apple",)
print(single_item, type(single_item))

not_a_tuple = ("apple")   # this is just a string!
print(not_a_tuple, type(not_a_tuple))

````

### Accessing Tuple Items

Tuples use the **same indexing and slicing rules** as lists.


````python
fruits = ("apple", "banana", "cherry")

print(fruits[0])    # first item
print(fruits[-1])   # last item
print(fruits[0:2])  # slicing works too!
````

### Why Use a Tuple if I Can't Change It?

- Tuples are great for data that should **never change** — like coordinates `(x, y)`, RGB colors `(255, 0, 0)`, or days of the week.
- Tuples are slightly **faster** than lists.
- Tuples can be used as **dictionary keys** (lists cannot!).


````python
# Trying to change a tuple gives an ERROR — this is expected!
fruits = ("apple", "banana", "cherry")
try:
    fruits[0] = "mango"
except TypeError as e:
    print("Error:", e)
````

````python
# Useful tuple operations
numbers = (4, 8, 1, 9, 3)

print("Length:", len(numbers))
print("Max:", max(numbers))
print("Min:", min(numbers))
print("Is 8 in tuple?", 8 in numbers)

# Converting between list and tuple
as_list = list(numbers)
as_list.append(100)          # lists CAN be changed
back_to_tuple = tuple(as_list)
print(back_to_tuple)
````

## 2️⃣ Sets

A **set** is a collection that:

- Has **no duplicate** values (duplicates are automatically removed!)
- Has **no fixed order** (you can't rely on item position)
- Is written with **curly brackets** `{ }`

Sets are perfect when you only care about **unique items**, not their order.


````python
# Creating a set
numbers = {1, 2, 3, 2, 1, 4}
print(numbers)   # duplicates are removed automatically!
print(type(numbers))
````

````python
# You CANNOT access set items by index (sets have no order)
numbers = {10, 20, 30}
try:
    print(numbers[0])
except TypeError as e:
    print("Error:", e)
````

### Modifying a Set


````python
fruits = {"apple", "banana", "cherry"}

fruits.add("mango")          # add a single item
print("After add:", fruits)

fruits.remove("banana")      # remove an item (error if not found)
print("After remove:", fruits)

fruits.discard("kiwi")       # remove if present, no error if missing
print("After discard (kiwi not present):", fruits)
````

### Set Operations (Math-style!)

Sets support handy math operations like **union**, **intersection**, and **difference**.


````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union (all items):        ", a | b)
print("Intersection (common):    ", a & b)
print("Difference (in a not b):  ", a - b)
print("Symmetric diff (not both):", a ^ b)
````

````python
# A very common use: removing duplicates from a list!
numbers_list = [1, 2, 2, 3, 3, 3, 4]
unique_numbers = list(set(numbers_list))
print(unique_numbers)
````

## 3️⃣ Dictionaries

A **dictionary** stores data as **key : value** pairs — like a real-world dictionary
where a *word* (key) maps to its *meaning* (value).

- Written with **curly brackets** `{ }`
- Format: `{key1: value1, key2: value2}`
- Keys must be **unique** and **immutable** (strings, numbers, or tuples)
- Values can be **anything**


````python
# Creating a dictionary
student = {
    "name": "Riya",
    "age": 21,
    "course": "Computer Science"
}
print(student)
print(type(student))
````

### Accessing Values

You access a dictionary's value using its **key**, not a numeric index.


````python
student = {"name": "Riya", "age": 21, "course": "Computer Science"}

print(student["name"])          # direct access
print(student.get("age"))       # safer access using .get()
print(student.get("grade", "Not Found"))   # default value if key is missing
````

### Modifying a Dictionary


````python
student = {"name": "Riya", "age": 21}

student["age"] = 22              # update an existing value
student["course"] = "AI"         # add a new key-value pair
print("After update/add:", student)

del student["course"]            # remove a key-value pair
print("After delete:", student)

student.pop("age")               # another way to remove
print("After pop:", student)
````

### Looping Through a Dictionary


````python
student = {"name": "Riya", "age": 21, "course": "Computer Science"}

# Loop through keys
for key in student:
    print(key)

print("---")

# Loop through keys and values together
for key, value in student.items():
    print(key, ":", value)
````

### Handy Dictionary Methods

| Method | What it does |
|---|---|
| `dict.keys()` | Returns all keys |
| `dict.values()` | Returns all values |
| `dict.items()` | Returns key-value pairs |
| `dict.get(key)` | Safely gets a value (no error if missing) |
| `key in dict` | Checks if a key exists |


````python
student = {"name": "Riya", "age": 21, "course": "Computer Science"}

print("Keys:  ", list(student.keys()))
print("Values:", list(student.values()))
print("Items: ", list(student.items()))
print("Is 'name' a key?", "name" in student)
````

### Bonus: Dictionary Comprehension

Just like list comprehension, you can build a dictionary in one line!

```
new_dict = { key_expr : value_expr for item in iterable }
```


````python
# Example: square numbers as a dictionary {number: square}
squares = {x: x**2 for x in range(1, 6)}
print(squares)
````

## 4️⃣ Quick Comparison — Which One Should I Use?

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Brackets | `[ ]` | `( )` | `{ }` | `{key: value}` |
| Ordered? | ✅ Yes | ✅ Yes | ❌ No | ✅ Yes (insertion order, Python 3.7+) |
| Changeable? | ✅ Yes | ❌ No | ✅ Yes | ✅ Yes |
| Duplicates allowed? | ✅ Yes | ✅ Yes | ❌ No | Keys: ❌ No, Values: ✅ Yes |
| Access by | Index | Index | Can't index | Key |
| Best for | General ordered data you'll change | Fixed/constant data | Unique items, membership tests | Labeled/structured data |
